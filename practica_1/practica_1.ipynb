# **PRÁCTICA 1: _Manejo básico de herramientas_**

## **_Objetivo_**
---

Que el alumno se familiarice con herramientas básicas utilizadas durante el desarrollo de proyectos de robótica: Ubuntu y su terminal, scripts de Bash, Jupyter, Markdown, Git y GitHub.

La práctica busca que el alumno pueda trabajar desde la terminal, automatizar tareas sencillas mediante scripts, documentar su trabajo en un cuaderno de Jupyter y compartir los resultados mediante un repositorio de GitHub.

## **_Introducción_**
---

En el desarrollo de proyectos de robótica es necesario utilizar diferentes herramientas de software para crear archivos, automatizar tareas, documentar procedimientos y compartir el trabajo. Aunque estas herramientas tienen funciones diferentes, su combinación permite establecer flujos de trabajo reproducibles y facilita la colaboración.

Ubuntu es un sistema operativo basado en Linux ampliamente utilizado en robótica, donde el manejo de la terminal permite ejecutar programas, administrar archivos y automatizar tareas sin depender exclusivamente de interfaces gráficas. Para ello se emplea Bash (*Bourne Again SHell*), un intérprete de comandos habitual en sistemas Linux que, además de permitir ejecutar comandos de manera interactiva, facilita la creación de *scripts* de Bash, es decir, archivos de texto que contienen una secuencia de comandos ejecutable de forma automatizada.

Para la integración y presentación del trabajo, Jupyter proporciona un entorno interactivo basado en *notebooks* estructurados por celdas que pueden contener texto, ecuaciones, código y resultados. En esta práctica, Jupyter se utilizará tanto para documentar el procedimiento como para ejecutar comandos de Bash desde el mismo documento. Para dar formato a este contenido, se emplea Markdown, un lenguaje de marcado ligero con una sintaxis sencilla que Jupyter utiliza para crear títulos, listas, tablas, enlaces y ecuaciones dentro de las celdas de texto.

Finalmente, el control y la gestión del código se realizan mediante Git y GitHub. Git es un sistema de control de versiones que permite registrar cambios en los archivos de un proyecto, recuperar versiones anteriores y trabajar mediante ramas. Por su parte, GitHub es una plataforma que permite alojar repositorios Git en línea y facilita el trabajo colaborativo a través de repositorios, ramas, colaboradores y herramientas de revisión.

En robótica, todas estas herramientas se utilizan conjuntamente de manera práctica. Por ejemplo, un alumno puede crear un *script* de Bash para preparar una carpeta de trabajo, documentar todo el procedimiento en Jupyter y, finalmente, almacenar tanto el *notebook* como los *scripts* en un repositorio de GitHub.

## **_Cuestionario Previo_**
---
•  **En Linux, ¿qué es una shell y qué relación existe entre sh y bash?**
> El shell es la capa más externa del sistema operativo. Los shells incorporan un lenguaje de programación para controlar procesos y archivos, además de iniciar y controlar otros programas. Este gestiona la interacción entre el usuario y el sistema operativo solicitándole la entrada, interpretando dicha entrada para el sistema operativo y gestionando cualquier resultado de salida procedente del sistema operativo. Los shells ofrecen un método para comunicarse con el sistema operativo. Esta comunicación tiene lugar de forma interactiva (la entrada desde el teclado se ejecuta inmediatamente) o como un script de shell. Un script de shell es una secuencia de mandatos del shell y del sistema operativo que se almacena en un archivo.

> - sh (Bourne Shell): Es el intérprete de comandos original de Unix creado por Stephen Bourne en 1977. En muchos sistemas modernos, sh es un enlace simbólico a otro shell más ligero que cumple estrictamente con el estándar POSIX (como dash en Ubuntu/Debian).

> - bash (Bourne-Again SHell): Fue creado en 1989 por Brian Fox para el proyecto GNU como un reemplazo libre y mejorado de sh. Conserva la compatibilidad con el sh original pero añade muchísimas funciones propias.

> Por otro lado el término "shell" se refiere a un programa que provee una interfaz de línea de comandos para interactuar con un sistema operativo. Bash (Bourne-Again SHell) es una de las terminales de Unix/Linux más comúnmente utilizados en muchas distribuciones de Linux. Aunque Bash es un tipo de shell, hay otras shells disponibles también, tales como la shell Korn (ksh), la shell C (csh), y la shell Z (zsh). Cada shell tiene su propia sintaxis y conjunto de características, pero todos comparten el propósito en común de proporcionar una interfaz de línea de comandos para interactuar con el sistema operativo.

•  **¿Qué es un script de Bash y cuál es la diferencia entre ejecutar un comando directamente en la terminal y ejecutarlo mediante un script?**
> Los Scripts Bash o Bash Scripts son simples archivos de texto que contienen listas de comandos que deben ser ejecutados en secuencia por el shell. Se escriben en archivos de texto plano e invocarlos es tan sencillo como referenciarlos en un comando de terminal. Bash es un intérprete de comandos o shell que actúa como una interfaz entre el usuario y el sistema operativo. Para que nos entendamos, Bash sería lo que conocemos como la interfaz de línea de comandos de los sistemas operativos. Aunque en la práctica Bash no es el único sistema de línea de comandos, es el más conocido y extendido. La inmensa mayoría de los sistemas Linux utilizan Bash como intérprete de línea de comandos. Los Bash Scripts están pensados para hacer todo tipo de tareas, aunque en la mayor parte de las ocasiones los administradores de sistemas los utilizan para la automatización de tareas repetitivas y complejas, simplificando nuestro día a día y aumentando la productividad.

> *1. Modo de interacción:*

> - *Terminal:* Escribes instrucción por instrucción y el intérprete (shell) responde en tiempo real.
> - *Script:* Escribes todo el código de antemano en un archivo (por ejemplo, con extensión .sh o .py) y el intérprete lo lee y ejecuta de arriba a abajo de forma secuencial.

> *2. Automatización y reutilización:*

> - *Terminal:* Si necesitas repetir la misma serie de 10 comandos mañana, tienes que teclearlos o buscarlos en el historial otra vez.
> -  *Script:* Guardas esos 10 comandos una vez y ejecutas el archivo completo con un solo comando.
	
> *3. Lógica de programación:*

> - _Terminal:_ Es ideal para consultas rápidas, pruebas puntuales o tareas simples.
> - *Script:*Permite añadir estructuras complejas como variables, condiciones (if/else), bucles (for/while) y manejo de errores.

> *4. Intervención humana:*

> - _Terminal:_ Requiere que estés presente y atento escribiendo en cada paso.
> - _Script:_ Puede ejecutarse en segundo plano, programarse con herramientas como cron o correr sin que el usuario esté supervisándolo.

•  **¿Qué es Jupyter y qué tipo de información puede contener una celda de un notebook?**
> Un Jupyter Notebook es una aplicación web de código abierto que permite a los científicos de datos crear y compartir documentos que incluyen código en vivo, ecuaciones y otros recursos multimedia.

> - Código: el notebook interpretará todo el contenido de una celda de código con el lenguaje Python. Al usar el código Python, al escribir ciertas cadenas, como from o el signo igual (=), el notebook les cambiará automáticamente el color o las resaltará para mayor claridad. Todas las líneas de código que empiecen por un signo de número (#) se interpretarán como comentarios, se pondrán en verde y en cursiva y no se ejecutarán como código en el notebook.
> - Markdown: el notebook interpretará todo el contenido de una celda Markdown con el lenguaje Markdown. Es un lenguaje sencillo para formatear texto enriquecido que utilizan en Internet clientes como GitHub. Consulte la Guía de Markdown en línea acerca del uso de Markdown. La ejecución de una celda Markdown convertirá su contenido en texto enriquecido. Cualquier línea que comience con uno o varios signos de número (#) se formateará como encabezado. También puede agregar código HTML sin formato a las celdas Markdown.
> - NBConvert sin formato: el notebook no procesará contenido de una celda NBConvert sin formato. Este tipo de celda se utiliza con poca frecuencia.

•  **¿Qué es Markdown y para qué se utiliza dentro de Jupyter?**
>  Markdown es un lenguaje de marcado que facilita la aplicación de formato a un texto empleando una serie de caracteres de una forma especial. En principio, fue pensado para elaborar textos cuyo destino iba a ser la web con más rapidez y sencillez que si estuviésemos empleando directamente HTML. Y si bien ese suele ser el mejor uso que podemos darle, también podemos emplearlo para cualquier tipo de texto, independientemente de cual vaya a ser su destino.

> • Anotaciones y explicaciones: Permite redactar introducciones, conclusiones o notas al pie que acompañan a los bloques de código.

> • Jerarquía y organización: Facilita la creación de títulos y subtítulos utilizando símbolos de numeral (#), dividiendo el cuaderno en secciones ordenadas (similar a un documento de texto).

> • Formato de texto: Permite aplicar negritas, cursivas, tachados, listas con viñetas o numeradas, y bloques de código de ejemplo.

> • Ecuaciones matemáticas: Soporta la escritura de fórmulas científicas y matemáticas utilizando sintaxis LaTeX.

> • Elementos multimedia: Permite insertar hipervínculos a páginas web e incorporar imágenes para ilustrar los resultados o el proceso.

> • Tablas y estructura visual: Ayuda a crear tablas de datos y líneas divisorias para mejorar la presentación visual del reporte.

•  **¿Cuál es la diferencia entre Git y GitHub?**

| Característica | Git | GitHub |
| :--- | :--- | :--- |
| **Naturaleza** | Un sistema de control de versiones** distribuido. | Una plataforma en la nube para alojar repositorios. |
| **Funcionamiento** | Funciona de manera local en tu computadora. | Funciona de manera remota (en servidores web). |
| **Conexión a internet** | No requiere internet para su uso diario (commits, ramas, etc.). | Requiere internet para la mayoría de sus funciones (subir, descargar, colaborar). |
| **Interfaz** | Principalmente se maneja por línea de comandos (terminal). | Cuenta con una interfaz gráfica web amigable. |
| **Colaboración** | Gestiona los cambios de forma individual en tu máquina. | Facilita la colaboración en equipo mediante *Pull Requests*, revisiones eissues. |
| **Alternativas** | Mercurio, SVN. | GitLab, Bitbucket. |




**3. GIT Y GITHUB**

Actividad visible en el repositorio

## **_Análisis de Resultados_**
---
¿Qué ventajas ofrece un script de Bash frente a ejecutar manualmente una secuencia de comandos en la terminal?
> Un script de Bash permite automatizar una secuencia de comandos y ejecutarla con una sola orden, en lugar de escribir cada línea a mano. Esto ahorra tiempo y reduce los errores humanos, porque el script siempre repite los mismos pasos en el mismo orden. Además se puede reutilizar en otras máquinas o por otras personas, y admite variables, condicionales, ciclos, funciones y argumentos, lo que lo vuelve flexible. También puede programarse para ejecutarse solo, por ejemplo con cron, y sirve como documentación de cómo se realizó un proceso, además de poder versionarse y mejorarse con el tiempo.


¿Qué diferencias existen entre una celda de Markdown y una celda de código en Jupyter?
> Una celda de Markdown contiene texto con formato, como títulos, listas, enlaces, imágenes o fórmulas en LaTeX, y se usa para explicar y organizar el notebook. Al ejecutarla solo se renderiza, es decir, se muestra el texto ya formateado, y no tiene numeración de entrada ni salida. En cambio, una celda de código contiene instrucciones del lenguaje del kernel, normalmente Python, y al ejecutarse el kernel las procesa y puede modificar variables y el estado del programa. Su resultado, que puede ser texto, tablas, gráficas o mensajes de error, aparece debajo de la celda y se identifica con la numeración In [n] y Out [n]. Combinar ambos tipos permite que la explicación y el código convivan en un mismo documento.


¿Qué utilidad tiene un repositorio en línea para almacenar scripts, notebooks y documentación de un proyecto de robótica?
> Un repositorio en línea es muy útil en un proyecto de robótica porque guarda el historial de cambios de scripts, notebooks y documentación, lo que permite volver a una versión anterior que funcionaba si algo falla. También funciona como respaldo centralizado, de modo que el trabajo no se pierde si se daña una computadora o un robot. Facilita la colaboración, ya que varios integrantes pueden trabajar al mismo tiempo mediante ramas y revisiones de código sin pisarse el trabajo. Además mantiene todo organizado en un solo lugar, con un README que explica el proyecto, permite que otras personas reproduzcan los experimentos clonando el repositorio, ayuda a llevar el seguimiento de errores y tareas pendientes, y sirve para compartir el proyecto y mostrarlo como evidencia de trabajo.


## **_Conclusiones_**
---
La realización de esta práctica cumplio con el objetivo de familiarizarse con las herramientas de software fundamentales para el desarrollo de proyectos de robótica . Primeramente, se logro el manejo de entornos y documentación, comprobando la utilidad de Jupyter y Markdown como herramientas para ejecutar comandos y fragmentos de código de forma interactiva y de igual manera estructurar reportes claros mediante el uso de texto enriquecido, tablas y expresiones matemáticas en LaTeX.
Por otra parte, mediante la Automatización con Bash, se comprendió la importancia de la terminal de Ubuntu y la creación de scripts de Bash. La automatización de tareas repetitivas, como la creación y despues limpieza de estructuras de directorios y archivos de texto, demostró ser una alternativa mucho más eficiente.
Finalmente, el control de versiones y colaboración mediante Git y GitHub y La implementación de un repositorio remoto (incluyendo archivos de configuración como gitignore, gestión de ramas y commits) facilita el trabajo en equipo.

## **_Bibliografía_**
---
Arsys. (s. f.). Bash Scripts: qué es y cómo funciona. https://www.arsys.es/blog/bash-scripts-que-es-y-como-funciona

Comandos esenciales del notebook—ArcGIS Notebook Server | Documentación de ArcGIS Enterprise. (s. f.). https://enterprise.arcgis.com/es/notebook/11.3/use/linux/essential-notebook-commands.htm

De Databricks, P. (2022, 25 febrero). ¿Qué es Jupyter Notebook? Databricks. https://www.databricks.com/es/blog/what-is-jupyter-notebook

Gomez, E. E. P. (2024, 23 julio). Tutorial de programación de Bash: Script de Shell de Linux y línea de comandos para principiantes. freeCodeCamp.org. https://www.freecodecamp.org/espanol/news/tutorial-de-programacion-de-bash-script-de-shell-de-linux-y-linea-de-comandos-para-principiantes/

Lasso, I. (2013, 18 septiembre). Qué es Markdown, para qué sirve y cómo usarlo. Genbeta. https://www.genbeta.com/guia-de-inicio/que-es-markdown-para-que-sirve-y-como-usarlo

Oluseye, J. (2024, 10 abril). Git frente a GitHub: Diferencias que todo desarrollador debe conocer. Datacamp. https://www.datacamp.com/es/blog/git-vs-github
